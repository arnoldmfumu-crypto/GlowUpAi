```python
import os

image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.webp', '.tiff'}
oily_path = 'oily'
dry_path = "dry"
normal_path = "normal"

count_1 = sum(
    1 for f in os.listdir(oily_path)
    if os.path.splitext(f)[1].lower() in image_extensions
)

print(f"Nombre d'images dans oily : {count_1}")

count_2 = sum(
    1 for f in os.listdir(dry_path)
    if os.path.splitext(f)[1].lower() in image_extensions
)

print(f"Nombre d'images dans dry : {count_2}")

count_3 = sum(
    1 for f in os.listdir(normal_path)
    if os.path.splitext(f)[1].lower() in image_extensions
)

print(f"Nombre d'images dans normal : {count_3}")
```
