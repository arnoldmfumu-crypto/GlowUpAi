# Nettoyage des données

### Suppression des colonnes inutiles

```python
df = df.drop(["Unnamed: 0", "Category", "Sold By", "Link", "Number of items"], axis=1)
```

```python
df
```

Plusieurs colonnes ici sont supprimées car elles n'ont soit pas d'intérêt pour l'entraînement du modèle ("Unnamed: 0"), soit présentent des valeurs incohérentes (1 seule valeur pour "Category"), soit n'apparaîtront pas dans les informations des produits communiquées à l'étape des résultats ("Sold By", "Link", "Number of items").

### Suppression des lignes inutiles

```python
df = df[~df["Product"].isin(["Nail paint", "Hand cream"])]
df
```

Les lignes dont les valeurs correspondent à "Nail paint" et "Hand cream" dans la colonne "Product" sont supprimées car elles ne font pas partie des produits proposés.

### Nettoyage des valeurs de la colonne "Skin type"

```python
df["Skin_Type"] = (
    df["Skin_Type"]
    .str.lower()
    .str.strip()
    .str.replace(r"<.*?>", "", regex=True)
)
```

```python
df
```

Un travail de structuration sur la colonne "Skin_Type" est nécessaire car les données avaient une forme et un contenu initialement très variés d'une ligne à l'autre.

### En résumé, ce dataset est riche et exploitable, mais un nettoyage ciblé permettra d’en tirer tout le potentiel pour l’analyse exploratoire ou l’ingestion dans un pipeline RAG.
