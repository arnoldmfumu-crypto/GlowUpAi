## Nettoyage

```python
# On droppe la colonne "Unnamed"

df_drop_col_unnamed = df.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
df
```

```python
# On exclut les lignes de produits de type "Manucure" et "Crème pour les mains" car pas concernées pour l'app

df = df[~df["Product"].isin(["Nail paint", "Hand cream"])]
df
```

```python
df["Product"].value_counts() # les lignes de produits "Hand cream" et "Nail paint" ont bien été masquées
```

```python
# On droppe la colonne "Category" car la valeur est la même sur toutes les lignes et n'apporte rien en termes d'infos
df.drop("Category", axis=1, inplace=True)
```

```python
df
```

```python
df["Brand"].value_counts()
```

```python
# On supprime la colonne "Prix"

df = df.drop("Price", axis=1)
```

```python
# la colonne "Number of items" n'est pas utile ici

df = df.drop("Number of items", axis=1)
```
