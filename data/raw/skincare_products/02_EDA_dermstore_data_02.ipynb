## Nettoyage des données

### Suppression des colonnes inutiles

```python
df.drop(["url","uniq_id", "scraped_at", "availability", "sku", "product_id", "price", "currency", "raw_ingredients", "raw_how_to_use", "raw_skin_type_and_concerns", "condition", "how_to_use", "range", "volume", "rating_value", "review_count"], axis=1, inplace=True)
```

Les valeurs cachées sont au nombre de 6 pour "ingredients" et 73 pour "skin_type_and_concerns".

```python
df["category"] = df["category"].str.split("/").str[-1].str.strip()
```

La colonne "category" présente des informations incohérentes son intitulé : les 2 premiers éléments, "Brands" et la marque séparés par des / de manière à ne laisser que le nom du produit.

```python
df
```

### Suppression des retours à la ligne et des espaces

```python
import re

def clean_text(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = re.sub(r"\s+", " ", text)  # espaces multiples
    return text.strip()

for col in df.columns:
    df[col] = df[col].apply(clean_text)
```

```python
df
```

### Supprimer le bruit marketing inutile

```python
def remove_noise(text):
    noise_patterns = [
        r"For the latest information.*",
        r"<em>.*?</em>"
    ]
    for pattern in noise_patterns:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE)
    return text
```

```python
df
```

### Nettoyage des ingrédients

```python
def clean_ingredients(text):
    if not text:
        return ""
    # enlever doublons simples
    ingredients = list(set([i.strip() for i in text.split(",")]))
    return ", ".join(ingredients)

df["ingredients"] = df["ingredients"].apply(clean_ingredients)
```

```python
df
```

### Gestion des valeurs manquantes

```python
df.fillna("", inplace=True)
```
