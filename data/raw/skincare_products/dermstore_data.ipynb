```python
# on fait la même chose pour le reste des colonnes concernées par le retrait des balises

df.columns = df.columns.map(lambda x: BeautifulSoup(str(x), "html.parser").get_text())
```

```python
# la colonne "category" présente des informations qui ne concerne pas son intitulé : on va supprimer les 2 premiers éléments qui sont "Brands" et ensuite la marque séparés par des /
# de cette façon il ne restera que le nom du produit
df["category"] = df["category"].str.split("/").str[-1].str.strip()
```
