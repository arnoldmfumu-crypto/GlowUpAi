### En résumé, ce dataset est riche et exploitable, mais il ne comporte pas d'informations directes qui répondent aux problématiques d'acné par exemple.
### Il faudra pour cela faire du prétraitement.

# Nettoyage des données

### Traitement préalable pour le RAG pour la problématique d'acné

```python
def generate_description(row):
    text = row["ingredients"].lower()
    
    desc = "This product is used for "

    if "salicylic" in text:
        desc += "treating acne and unclogging pores, "
    if "hyaluronic" in text:
        desc += "hydration, "
    if "retinol" in text:
        desc += "anti-aging and skin renewal, "
    if "niacinamide" in text:
        desc += "reducing inflammation and oil control, "

    return desc.strip(", ")

df["generated_description"] = df.apply(generate_description, axis=1)
```
